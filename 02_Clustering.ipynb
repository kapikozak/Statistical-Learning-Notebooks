### Kullback-Leibler divergence

For random vectors $U, V \in \mathbb R^n$ valued in finite set $\{\beta_1, \dots, \beta_N\}$ with $\mathbb P(U = \beta_i) = p_i$ and $\mathbb P(V = \beta_i) = q_i$, the Kullback-Leibler divergence is defined as
$$
\operatorname{D}(U \mid V) = \sum_{i = 1}^N p_i\log\Bigl(\frac{p_i}{q_i}\Bigr).
$$

This value is always non-negative and equal to zero if and only if $\forall i p_i = q_i$. This follows from the inequality $\forall x > 0 \log x \ge 1 - 1/x$, with equality only if $x = 1$. From that,
$$
\operatorname{D}(U \mid V) \ge \sum_{i = 1}^N p_i - \sum_{i = 1}^N q_i = 0.
$$

### EM Clustering for Gaussian Mixtures

Observed Data: $X \in \mathbb R^{n \times p}$ matrix containing independent random vectors $x_1, \dots, x_n \in \mathbb R^p$ in rows.

Hidden Variable: $Z \in \mathbb R^n$ random vector with identically distributed independent components $z_i \in \{1, \dots, k\}$. We furthermore assume that $((x_1, z_1), \dots, (x_n, z_n))$ has independent components.

Model Parameters: $\theta = (\pi_1, \dots, \pi_k, \mu_1, \dots, \mu_k, \Sigma_1, \dots, \Sigma_k)$ mixture weights, means and covariances of Gaussian components.

Let $U \sim \mu_{Z \mid X}$ under $\theta^{(old)}$ and $V \sim \mu_{Z \mid X}$ under $\theta$. From Bayes theorem it follows that
$$
  p_\theta(x) = \frac{p_\theta(x, \alpha)}{p_{\theta}(\alpha \mid x)}.
$$

We can rewrite
\begin{align*}
  l(x \mid \theta)
  &= \sum_\alpha p_{\theta^{(old)}}(\alpha \mid x)\log p_\theta(x) \\
  &= \sum_\alpha p_{\theta^{(old)}}(\alpha \mid x)\log \Bigl(\frac{p_\theta(x, \alpha)}{p_\theta(\alpha \mid x)}\Bigr)\\
  &= \sum_\alpha p_{\theta^{(old)}}(\alpha \mid x) \log \Bigl(\frac{p_\theta(x, \alpha)}{p_{\theta^{(old)}}(\alpha \mid x)}\Bigr)\\
  &\quad + \sum_\alpha p_{\theta^{(old)}}(\alpha \mid x) \log\Bigl(\frac{p_{\theta^{(old)}}(\alpha \mid x)}{p_\theta(\alpha \mid x)}\Bigr)\\
  &= \sum_\alpha p_{\theta^{(old)}}(\alpha \mid x) \log \Bigl(\frac{p_\theta(x, \alpha)}{p_{\theta^{(old)}}(\alpha \mid x)}\Bigr) + \operatorname{D}(U \mid V).
\end{align*}

As $D(U \mid V) \ge 0$ with equality if and only if $\forall \alpha\, p_\theta(\alpha \mid x) = p_{\theta^{(old)}}(\alpha \mid x)$, then in particular
\begin{align*}
  l(x \mid \theta)
  &\ge \sum_\alpha p_{\theta^{(old)}}(\alpha \mid x) \log \Bigl(\frac{p_\theta(x, \alpha)}{p_{\theta^{(old)}}(\alpha \mid x)}\Bigr)\\
  &= \sum_\alpha p_{\theta^{(old)}}(\alpha \mid x) (\log p_\theta(x, \alpha) - \log p_{\theta^{(old)}}(\alpha \mid x))
\end{align*}
with equality if and only if $\forall \alpha\, p_\theta(\alpha \mid x) = p_{\theta^{(old)}}(\alpha \mid x)$.

The second part of the term does not depend on $\theta$, so by taking
$$
  \theta^{(new)} = \operatorname{argmin}_\theta -\sum_\alpha p_{\theta^{(old)}}(\alpha \mid x) \log p_\theta(x, \alpha),
$$
we achieve
\begin{align*}
  l(x \mid \theta^{(new)})
  &\ge \sum_\alpha p_{\theta^{(old)}}(\alpha \mid x) \log \Bigl(\frac{p_{\theta^{(new)}}(x, \alpha)}{p_{\theta^{(old)}}(\alpha \mid x)}\Bigr)\\
  &\ge \sum_\alpha p_{\theta^{(old)}}(\alpha \mid x) \log \Bigl(\frac{p_{\theta^{(old)}}(x, \alpha)}{p_{\theta^{(old)}}(\alpha \mid x)}\Bigr) = l(x \mid \theta^{(old)}).
\end{align*}

### E - step

We compute posterior responses $(\gamma_{i, j})_{i = 1, \dots, n,\, j = 1, \dots, k}$,
$$
  \gamma_{i, j} = p_{\theta^{(old)}}(j \mid x_i) = \frac{\pi_j^{(old)}p_{\theta^{(old)}}(x_i \mid j)}{\sum_{l = 1}^k \pi_l^{(old)} p_{\theta^{(old)}}(x_i \mid l)} = \boxed{\frac{\pi_j^{(old)}f_{\mathrm N(\mu_j^{(old)}, \Sigma_j^{(old)})}(x_i)}{\sum_{l = 1}^k\pi_l^{(old)}f_{\mathrm N(\mu_l^{(old)}, \Sigma_l^{(old)})}(x_i)}}.
$$

Then
$$
  p_{\theta^{(old)}}(\alpha \mid x) = \prod_{m = 1}^n p_{\theta^{(old)}}(\alpha_m \mid x_m) = \prod_{m = 1}^n\gamma_{m, \alpha_m}.
$$

### M  - step

We minimise
$$
  -\sum_\alpha p_{\theta^{(old)}}(\alpha \mid x) \log p_\theta(x, \alpha) \to \min_\theta.
$$

For any fixed $i$,
$$
  \prod_{m \ne i}\sum_{r = 1}^k\gamma_{m, r} = \prod_{m \ne i} 1 = 1.
$$

Then
\begin{align*}
  &-\sum_\alpha p_{\theta^{(old)}}(\alpha \mid x) \log p_\theta(x, \alpha)\\
  &\quad = -\sum_\alpha \prod_{m = 1}^n\gamma_{m, \alpha_m} \sum_{i = 1}^n(\log \pi_{\alpha_i} + \log p_\theta(x_i \mid \alpha_i))\\
  &\quad = -\sum_{i = 1}^n \sum_{j = 1}^k \gamma_{i, j}(\log \pi_j + \log p_\theta(x_i \mid j))\sum_{\alpha : \alpha_i = j}\prod_{m \ne i}\gamma_{m, \alpha_m}\\
  &\quad = -\sum_{i = 1}^n \sum_{j = 1}^k \gamma_{i, j}(\log \pi_j + \log p_\theta(x_i \mid j))\prod_{m \ne i}\sum_{r = 1}^k\gamma_{m, r}\\
  &\quad = -\sum_{i = 1}^n \sum_{j = 1}^k \gamma_{i, j}(\log \pi_j + \log p_\theta(x_i \mid j)).
\end{align*}

As $p_\theta(x_i \mid j) = f_{\mathrm N(\mu_j, \Sigma_j)}(x_i)$, the last term is
$$
  \propto \sum_{i = 1}^n \sum_{j = 1}^k\gamma_{i, j}\Bigl(-\log \pi_j + \frac{1}{2}\log |\Sigma_j| + \frac{1}{2}(x_i - \mu_j)^\top\Sigma_j^{-1}(x_i - \mu_j)\Bigr) := Q(\theta \mid \theta^{(old)}).
$$

### Minimisation

Let denote $\Gamma_j = \gamma_{1j} + \dots + \gamma_{nj}$,
$$
  f(\pi) = -\sum_{j = 1}^k\Gamma_j\log \pi_j
$$
and $L(\pi, \lambda) = f(\pi) + \lambda(\pi_1 + \dots + \pi_k - 1)$.

We have
$$
  \frac{\partial L}{\partial \pi_j}(\pi) = -\frac{\Gamma_j}{\pi_j} + \lambda,
$$
which means $D_\pi L(\pi) = 0$ if and only if $\forall j\, \pi_j = \Gamma_j/\lambda$. Since $\pi_1 + \dots + \pi_k = 1$, we obtain $\lambda = n$.

Matrix $D^2L(\pi)$ is diagonal with terms $\Gamma_j/\pi_j^2$. In particular, $D^2L(\pi)$ is positive definite for all $\pi$ such that $\forall j\, \pi_j > 0$. From that,
$$
  \hat \pi_j = \boxed{\frac{\Gamma_j}{n}}.
$$

While covariances $\Sigma_1, \dots, \Sigma_k$ are fixed, let
$$
  g(\mu_j) = \sum_{i = 1}^n \frac{1}{2}\gamma_{i,j}(x_i - \mu_j)^\top\Sigma_j^{-1}(x_i - \mu_j).
$$

Then
$$
  Dg(\mu_j) = \Sigma_j^{-1} \Bigl(\Gamma_j\mu_j - \sum_{i = 1}^n \gamma_{i,j}x_i\Bigr).
$$

By setting $Dg(\mu_j) = 0$, we obtain
$$
  \mu_j = \frac{\sum_{i = 1}^n \gamma_{i,j}x_i}{\Gamma_j}.
$$

As $\forall \mu_j\, D^2g(\mu_j) = \Gamma_j\Sigma_j^{-1}$, where $\Gamma_j >0$ and $\Sigma_j^{-1}$ is symmetric positive definite, then the point above is a global minimum. We conclude that
$$
  \hat\mu_j = \boxed{\frac{\sum_{i = 1}^n \gamma_{i,j}x_i}{\Gamma_j}}.
$$

For every non-singular covariance matrix $\Sigma_j$ there exist real positive numbers $d_1, \dots, d_p$ and orthogonal matrix $U$, such that $\Sigma_j = U \operatorname{diag}(d_1, \dots, d_p) U^\top$.

Using this decomposition,
\begin{align*}
  (x_i - \mu_j)^\top\Sigma_j^{-1}(x_i - \mu_j)
  &= (x_i - \mu_j)^\top U\operatorname{diag}(1/d_1, \dots, 1/d_p)U^\top (x_i - \mu_j)\\
  &= \operatorname{tr}(x_i - \mu_j)^\top U\operatorname{diag}(1/d_1, \dots, 1/d_p)U^\top (x_i - \mu_j)\\
  &= \operatorname{tr}\operatorname{diag}(1/d_1, \dots, 1/d_p)U^\top (x_i - \mu_j)(x_i - \mu_j)^\top U\\
  & = \sum_{l = 1}^p \frac{1}{d_l}(U^\top (x_i - \mu_j)(x_i - \mu_j)^\top U)_{ll}.
\end{align*}

Denoting
$$
  S_j = \sum_{i = 1}^n\gamma_{ij} (x_i - \mu_j)(x_i - \mu_j)^\top,
$$
we can rewrite
$$
  \sum_{i = 1}^n\gamma_{ij}\Bigl(\frac{1}{2}\log |\Sigma_j| + \frac{1}{2}(x_i - \mu_j)^\top\Sigma_j^{-1}(x_i - \mu_j)\Bigr) = \frac{1}{2}\sum_{l = 1}^p \Bigl(\Gamma_j\log d_l + \frac{(U^\top S_j U)_{ll}}{d_l}\Bigr).
$$

While matrix $U$ is fixed we can see the above as function $h$ of variable $d$. In that case
$$
  \frac{\partial h}{\partial d_l}(d) = \frac{\Gamma_j}{2d_l} - \frac{(U^\top S_j U)_{ll}}{2d_l^2}.
$$

We conclude that $Dh(d) = 0$ if and only if
$$
  \forall l \quad d_l = \frac{(U^\top S_j U)_{ll}}{\Gamma_j} := \delta_l.
$$

Differentiating second time,
$$
  \frac{\partial^2 h}{\partial d_l^2}(d) = -\frac{\Gamma_j}{2d_l^2} + \frac{(U^\top S_j U)_{ll}}{d_l^3}
$$
and
$$
  \forall r \ne l \quad \frac{\partial^2 h}{\partial d_r \partial d_l}(d) = 0.
$$

Assuming that $S_j$ is positive definite, matrix $D^2h(\delta_l)$ is diagonal with terms
$$
  -\frac{\Gamma_j^3}{2(U^\top S_j U)_{ll}^2} + \frac{\Gamma_j^3}{(U^\top S_j U)_{ll}^2} = \frac{\Gamma_j^3}{2(U^\top S_j U)_{ll}^2} > 0.
$$

Hence we conclude that $\delta$ is a local minimum of function $h$.

Under the same assumption,
$$
  \frac{1}{2}\Bigl(\Gamma_j\log d_l + \frac{(U^\top S_j U)_{ll}}{d_l}\Bigr) = \frac{\Gamma_jd_l\log d_l + (U^\top S_j U)_{ll}}{2d_l} \to \infty
$$
both when $d_l \to 0^+$ and $d_l \to \infty$. Therefeore  for $S_j$ positive definite $\delta$ is a global minimum of function $h$.

Let
\begin{align*}
  v(U)
  &= \frac{1}{2}\sum_{l = 1}^p \Bigl(\Gamma_j\log \delta_l + \frac{(U^\top S_j U)_{ll}}{\delta_l}\Bigr)\\
  & = \frac{\Gamma_j}{2}\sum_{l = 1}^p \log(U^\top S_j U)_{ll}  + \frac{p}{2}\Gamma_j(1 - \log \Gamma_j) = \frac{\Gamma_j}{2} \log \prod_{l = 1}^p(U^\top S_j U)_{ll} + C.
\end{align*}

Using Hadamard's inequality,
$$
  v(U) \ge \frac{\Gamma_j}{2} \log |U^\top S_j U| + C = \frac{\Gamma_j}{2} \log |S_j| + C.
$$

Equality holds if and only if $U^\top S_j U$ is diagonal, in which case
$$
  \Sigma_j = U \operatorname{diag}(d_1, \dots, d_p) U^\top = \frac{S_j}{\Gamma_j}.
$$

Hence
$$
  \hat \Sigma_j = \boxed{\frac{S_j}{\Gamma_j}}.
$$

### Likelihood transition revisit nd comprassion with K-means

As we have seen before, the log-likelihood may increase even if the M-step does not maximize $Q(\theta \mid \theta^{(old)})$. Indeed, a decrease in $Q$ can be compensated by the positive KL-divergence term arising from the change in the posterior distribution of the latent variables. Thus, even when $Q(\theta^{(new)} \mid \theta^{(old)}) < Q(\theta^{(old)} \mid \theta^{(old)})$ the log-likelihood can still increase if the corresponding Kullback-Leibler divergence is sufficiently large.

In K-means we have hard assignments and minimise an objective function of the form
$$
  -\frac{1}{n}\sum_{i = 1}^n \sum_{j = 1}^k \operatorname 1(i \in C_j)\log e^{-|x_i - \mu_{C_j}|^2}.
$$

Thus, K-means can be viewed as a special case of a Gaussian mixture model with hard assignments and fixed covariance matrices. Gaussian mixture models therefore provide a more flexible framework for fitting the data, with EM providing a natural algorithm for estimating their parameters.
